```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar los datos
df = pd.read_csv('munsell-array-fixed-choice.csv')

def plot_color_distribution(language_column, title):
    # 2. Calcular porcentajes
    # Contamos cuántas veces aparece cada color y dividimos por el total
    color_counts = df[language_column].value_counts(normalize=True) * 100

    # Convertimos a DataFrame para graficar más fácil
    data = color_counts.reset_index()
    data.columns = ['ColorName', 'Percentage']

    # 3. Crear la gráfica
    plt.figure(figsize=(10, 6))

    # Creamos las barras horizontales
    # Usamos el nombre del color para intentar darle ese mismo color a la barra
    bars = plt.barh(data['ColorName'], data['Percentage'], color='skyblue', edgecolor='black')

    # Opcional: Intentar que el color de la barra coincida con el nombre (si existe en CSS)
    for i, bar in enumerate(bars):
        color_name = data['ColorName'].iloc[i]
        try:
            bar.set_color(color_name)
            bar.set_edgecolor('black')
        except:
            continue # Si el nombre no es un color válido de CSS, deja el default

    # Configuración de estética
    plt.title(title, fontsize=14)
    plt.xlabel('Porcentaje de chips (%)', fontsize=12)
    plt.ylabel('Nombre del color', fontsize=12)
    plt.gca().invert_yaxis()  # Para que el color más frecuente aparezca arriba
    plt.grid(axis='x', linestyle='--', alpha=0.7)

    plt.tight_layout()
    plt.show()

# Ejecutar para inglés (puedes cambiarlo por 'spanish_color' o 'tsimane_color')
plot_color_distribution('english_color', 'Distribución de Nombres de Colores (Inglés)')
```

# Naming colors

If you are shown random colors evenly spaced across the color spectrum, what percentage of those colors would the average person name red? What about green?

English speakers in the United States were shown 80 color chips in random order in controlled lighting condtions. These 80 colors are evenly spaced across the standard Munsell array of colors. The participants were told: "There are 11 choices: black, white, red, green, blue, purple, brown, yellow, orange, pink, gray. Choose the closest color word".

This study was then repeated with Bolivian-Spanish speakers in Bolivia and Tsimane' speakers from the Amazon.

The dataset `munsell-array-fixed-choice.csv` shows the most common name in each language for each color chip.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("munsell-array-fixed-choice.csv")
df.head()
```

The function below plots the 80 evenly spaced colors that each study participant was shown in random order.

```python
def plot_colors(color_column):
    fig, ax = plt.subplots()
    for idx, row in df.iterrows():
        ax.add_patch(plt.Rectangle((row['x'] - 0.5, row['y'] - 0.5), 1, 1, color=row[color_column], linewidth=0))

    ax.set_xlim(0.5, 20.5)
    ax.set_ylim(0.5, 8.5)
    ax.set_aspect('equal')
    ax.set_facecolor('black')
    ax.set_xticks(range(1, 21))
    ax.set_yticks(range(1, 9), ['H', 'G', 'F', 'E', 'D', 'C', 'B', 'A'])

plot_colors('tile_hex')
```

We can also use this function to plot the most common name in each language for each color chip.

```python
# Also try 'spanish_color' and 'tsimane_color'
plot_colors('english_color')
```
